왜 데이터 마이닝을 하는가? 

- 현재 서비스 -> 효율 중심 (교통상황 , 거리등의 데이터로 연산하여 경로 생성)

- 모먼트 루트 -> 경험중심 (어떤 데이터(기준)로 경로를 생성할지 결정 못함.)


경로 생성 알고리즘 자체보다 "무엇을 기준으로 삼을지"를 먼저 정의하는 게 핵심 난제이고, 그 정의 과정에서 데이터 마이닝(연관 규칙 등)을 쓰겠다.

#### 연관 규칙(Association Rule)
데이터간의 연관성과 숨겨진 패턴을 찾아내는 기법

무엇을 찾을지 결정을 못했을때 적용할수있음.
무엇을 어떻게 찾을지는 이미 정해진 방법(알고리즘)을 적용함.

정리
1. 분석 방법(how to search) = 정해져 있음 (알고리즘)
2. 분석 결과/타겟(what the answer is) = 모름 (이게 탐색의 목적)
